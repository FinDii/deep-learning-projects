```python
import os
import random
import json
from matplotlib import pyplot as plt
from PIL import Image, ImageDraw, ImageFont
```

```python
# TODO: replace with your D2-City test image directory
test_img_dir = "/kaggle/input/datasets/xiaoyuwan/didi-detection/dataset/dataset_release/test"
# Images are saved here
output_dir = "/kaggle/working/data/mllm_detection/images"
os.makedirs(output_dir, exist_ok=True)

# Note on Generate. Refer to the archived source for the detailed original comment.
# 由于是第二次跑，就直接指定了
indices = ["11921", "11722", "11486", "11251", "11225"]
if not indices:
    # Remove all images in output_dir
    for file in os.listdir(output_dir):
        if file.endswith((".jpg", ".png")):
            os.remove(os.path.join(output_dir, file))
    # Random 5 images
    indices = random.sample(range(11000, 12000), 5)

test_img_files = [os.path.join(test_img_dir, f"{index}.jpg") for index in indices]

print("选取的图像索引:", ", ".join(map(lambda x: f"{x}.jpg", indices)))
print("\n图像路径:")

# Note on Image. Refer to the archived source for the detailed original comment.
copied_files = []
for i, src_path in enumerate(test_img_files):
    if os.path.exists(src_path):
        dst_path = os.path.join(output_dir, f"{indices[i]}.jpg")
        img = Image.open(src_path)
        # Note on Image, Input. Refer to the archived source for the detailed original comment.
        # Center crop a square part and resize to 1024x1024
        width, height = img.size
        x1 = width // 2 - height // 2
        img = img.crop((x1, 0, x1 + height, height))
        img = img.resize((1024, 1024))
        img.save(dst_path)
        copied_files.append(dst_path)
        print(f"  {i+1}. {src_path}")
    else:
        print(f"  {i+1}. {src_path} (不存在)")

print(f"\n共复制 {len(copied_files)} 张图像到 {output_dir}")
```

### Results / Visualization / Image

This cell documents the traffic scene object detection experiment. For the full original methodology, mathematical derivations, and commentary, see the archived notebook in `archive/original_notebooks/`.


```python
import matplotlib.patches as patches
import numpy as np

# ---------- 全局配置 ----------
CATEGORY_COLORS = [
    (255, 0, 0),
    (0, 255, 0),
    (0, 0, 255),
    (255, 255, 0),
    (255, 0, 255),
    (0, 255, 255),
    (128, 0, 128),
    (0, 128, 128),
    (128, 128, 0),
    (128, 128, 128),
    (64, 64, 64),
    (192, 192, 192),
]

category_names = [
    "car", "van", "bus", "truck", "person", "bicycle",
    "motorcycle", "open-tricycle", "closed-tricycle",
    "forklift", "large-block", "small-block"
]

def visualize_detection(image_path, annotations, title="MLLM Detection"):
    """
    使用 Matplotlib 绘制检测框和标签（自动判断坐标类型）
    """
    img = plt.imread(image_path)
    fig, ax = plt.subplots(1, figsize=(10, 10))
    ax.imshow(img)

    h, w = img.shape[:2]

    for ann in annotations:
        bbox = ann.get("bbox", [0, 0, 0, 0])
        # 判断是否归一化坐标
        if all(0 <= v <= 1 for v in bbox):
            x, y, bw, bh = bbox
            x *= w
            y *= h
            bw *= w
            bh *= h
        else:
            x, y, bw, bh = bbox

        category = ann.get("category_id", -1)
        if category < 0 or category >= len(CATEGORY_COLORS):
            color = 'gray'
            label = "Unknown"
        else:
            color = np.array(CATEGORY_COLORS[category]) / 255.0  # 转为 0-1
            label = category_names[category] if category < len(category_names) else "Unknown"

        # Note on Plot. Refer to the archived source for the detailed original comment.
        rect = patches.Rectangle((x, y), bw, bh, linewidth=2, edgecolor=color, facecolor='none')
        ax.add_patch(rect)

        # 添加文字标签（带背景框）
        ax.text(x, y - 5, label, color='white', fontsize=12,
                bbox=dict(facecolor=color, alpha=0.6, pad=2))

    ax.axis('off')
    return fig

# Note on Results, Visualization, Define. Refer to the archived source for the detailed original comment.
# 例如：
# mllm_results = { "11225": {"annotations": [{"category_id": 0, "bbox": [0.1, 0.2, 0.3, 0.4]}]} }
# output_dir = "/kaggle/working/images"

if 'mllm_results' in globals() and mllm_results:
    num_images = len(mllm_results)
    fig, axes = plt.subplots(num_images, 1, figsize=(8, 8 * num_images))
    if num_images == 1:
        axes = [axes]

    for i, (img_id, result) in enumerate(mllm_results.items()):
        img_path = os.path.join(output_dir, f"{img_id}.jpg")
        if os.path.exists(img_path):
            annotations = result.get("annotations", [])
            ax = axes[i]
            img = plt.imread(img_path)
            ax.imshow(img)
            h, w = img.shape[:2]
            for ann in annotations:
                bbox = ann.get("bbox", [0, 0, 0, 0])
                if all(0 <= v <= 1 for v in bbox):
                    x, y, bw, bh = bbox
                    x *= w
                    y *= h
                    bw *= w
                    bh *= h
                else:
                    x, y, bw, bh = bbox

                category = ann.get("category_id", -1)
                color = np.array(CATEGORY_COLORS[category]) / 255.0 if 0 <= category < len(CATEGORY_COLORS) else 'gray'
                label = category_names[category] if 0 <= category < len(category_names) else "Unknown"
                rect = patches.Rectangle((x, y), bw, bh, linewidth=2, edgecolor=color, facecolor='none')
                ax.add_patch(rect)
                ax.text(x, y - 5, label, color='white', fontsize=12,
                        bbox=dict(facecolor=color, alpha=0.6, pad=2))
            ax.set_title(f"Image {img_id}\n{len(annotations)} objects")
            ax.axis('off')
        else:
            print(f"错误！图像 {img_path} 不存在！")

    plt.tight_layout()
    plt.show()
else:
    print("请先将 MLLM 输出结果粘贴到代码中！")
```
